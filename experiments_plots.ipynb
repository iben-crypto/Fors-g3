# Combined Experiments (1-5)\nThis notebook reads `experiment_1.csv` to `experiment_5.csv` and plots force and wheel channels using aligned time.\n

```python
import csv\nfrom pathlib import Path\nimport matplotlib.pyplot as plt\n\nrepo = Path('.')\nfiles = [repo / f'experiment_{i}.csv' for i in range(1, 6)]\n\ndef read_experiment(path):\n    with path.open(newline='') as f:\n        rows = list(csv.DictReader(f))\n    t = [float(r['time_aligned']) for r in rows]\n    force = [float(r['force_cal_0']) for r in rows]\n    c0 = [float(r['wheel_cal_0']) for r in rows]\n    c1 = [float(r['wheel_cal_1']) for r in rows]\n    c2 = [float(r['wheel_cal_2']) for r in rows]\n    return t, force, c0, c1, c2\n
```

```python
for i, path in enumerate(files, start=1):\n    t, force, c0, c1, c2 = read_experiment(path)\n    fig, axes = plt.subplots(2, 1, figsize=(12, 8), sharex=True)\n    axes[0].plot(t, force, label='Force cal[0]', color='tab:blue')\n    axes[0].set_title(f'Experiment {i} - Force')\n    axes[0].set_ylabel('Force')\n    axes[0].grid(True, alpha=0.3)\n    axes[0].legend()\n    axes[1].plot(t, c0, label='Wheel cal[0]')\n    axes[1].plot(t, c1, label='Wheel cal[1]')\n    axes[1].plot(t, c2, label='Wheel cal[2] (acceleration)')\n    axes[1].set_title(f'Experiment {i} - Wheel channels')\n    axes[1].set_xlabel('Aligned time [s]')\n    axes[1].set_ylabel('Wheel values')\n    axes[1].grid(True, alpha=0.3)\n    axes[1].legend()\n    plt.tight_layout()\n    plt.show()\n
```
